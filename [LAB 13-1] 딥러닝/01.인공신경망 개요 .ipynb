# [LAB 12] 딥러닝 > 신경망의 이해 > 1. 인공신경망 개요

## 01. 인공신경망의 이해
- 여러개의 입력을 받아 계산하고 출력을 만들어 낸다 (독립변수에 해당하는 피쳐를 입력하여 타겟, 종속변수를 출력)
- 입력값 미가공 데이터 또는 다른 출력 값이 될 수 있다
- 출력신호는 문제의 최종적인 해가 되거나 다른 뉴런에 입력이 될 수 있다
- 뉴런, 종속변수가 또 다른 독립변수처럼 작용 가능



## #02.인공신경망의 구성 요소


### [1] 가중치와 편향
#### (1) 가중치의 이해 --> 파티에 가기 위한 조건 3가지
예를 들어 파티 입장권이 비싸면? 날씨나 친구의 동행 여부 보다 알바비 입금 여부가 파티에 갈 수 있는지 결정하는 더 중요한 요인이 된다

결국 파티의 참석과 불참석을 결정하기 위한 조건값들은 일정한 가중치가 곱해져야 공평한 조건 비교가 된다

날씨의 중요도 : 2
친구 동행 여부의 중요도 : 3
알바비의 중요도 : 10


[예시]
날씨 ㅇ , 친구 ㅇ , 알바비 x --> 파티 못감
날씨  x , 친구 x  , 알바비 ㅇ--> 파티 감

*여기서의 가중치는? 계수같은 역할


#### (2) 편향
- 만약 이 사람이, 성향이 E 라면? 어떤 요인이 있어도 가려고 할거고 I 라면 돈이 있어도 안가겠지?
- 파티에 참석하는 것을 좋아하는 성향이 있는 A 와 싫어하는 성향을 가진 B 가 있을 떄 이들의 3가지 요인 (조건) 과 중요도가 완전히 동일하다고 가정
- ax + b : a 는 가중치, b 는 편향
  

- A 의 경우 (파티 좋아하는 E)
  --> 날씨 ㅇ , 친구 ㅇ , 알바비 x --- 파티좋아하는 성향---> 파티 감

- B 의 경우 (파티 좋아하는 E)
  --> 날씨 ㅇ , 친구 ㅇ , 알바비 ㅇ --- 파티 안 좋아하는 성향---> 파티 안 감



  #### (3) 정리
  - 인공신경망의 결과값 (파티 참석 여부) 로 출력되는 것은 일종의 의사결정 과정
  - 그 결과는 각 요인에 곱해지는 가중치 (계수) 와 편향 (절편) 에 의해 결정된다
  - 머신 러닝의 궁극적 목표는 겨룩 여러 독립변수들로부터 하나 혹은 그 이상의 종속변수에 도달하는데 필요한 가중치와 편향을 찾는 것


## #03. 뉴련과 활성화 함수의 이해
### (1) 뉴런
- 위의 계산을 진행하는 과정을 뉴런이라 하고, 이 계산에 적용되는 것이 활성화 함수
- ax+b 가 있을 때, a와 b 를 찾듯이 0~1 사이의 숫자가 뉴런
- 숫자들의 조합에 따라 뉴런이 결정된다
- 화살표 실선의 역할은 특정 실수 (real number) 의미
- 이 실수는 + 값일수도 있고 - 값일 수 도 있다



0.4---->
0.3---->    ?
0.9---->


#### 가중치를 어떤 규칙도 없이 임의로 결정한 경우
- 뉴런의 값과 가중치의 값을 곱한 결과를 모두 합한 결과에 임의의 값 편향을 더한 결과가 ? 의 값이 된다
- wx+b = y (원점을 지나면? b 는 없을 수도 있음)
- 편향은 있을 수도 있고, 없을 수도 있다



0.4---->  [-5.3]
0.3---->  [1.2]    ?
0.9---->  [0.8] 

이렇게 각자 다른 함수 (활성화 함수) 들을 거쳐서 공통된 값에 도달하는 것

? 의 결과값이 (-2.12 + 0.36 + 0.72) = 2.3 이라고 한다면
뉴런은 0~1 사이의 숫자가 나와야하는 것과 위배된다
따라서 이 결과값은 특정 처리를 거쳐 0-1 사이의 값으로 변환되어야 한다


### (2) 활성화 함수를 통한 처리
- 가중치나 편향은 특별한 제약 없이 모든 Real Number 가 될 수 있으므로 무한대에 가까운 양수나 실수가 될 수도 있다
- 그러므로 계산된 결과값이 0-1 사이의 값이라는 조건에 위배되기 쉽다
- 활성화 함수는 계산한 결과값을 파라미터로 전달해서 0-1 사이의 값으로 변환하여 리턴한다
- 활성화 함수는 파라미터 값을 변환한 후에도 상대적인 크기가 유지되어야 한다
- a,b,c 가 크기 순서대로 a>b>c 라고 한다면, a,b,c 를 활성화 함수에 전달하여 반환 받은 결과도 이러한 크기 순서가 유지되어야 한다
- 즉 활성화 함수는 모든 x 에 대해 항상 우상향이다
- 딥러닝은 실수 연산이 전제 (정수 x)
- 실수 연산 제대로 못하는건 컴퓨터 CPU (0.1 + 0.2 ==0.3 은 FALSE)
- 실수 이진수 연산에 더 최적화 되어 있는건 GPU
  

### (3) 뉴런의 계산
- 뉴런은 활성화 함수를 이용해 출력되는 값
- 입력 신호의 가중치 합을 계산하여 임계값과 비교
- 가중치 합이 임계값보다 작으면 뉴런의 출력은 -1, 같거나 크면 +1 출력
- X 는 뉴런으로 들어가는 입력의 순 가중합
- w 는 입력 i 가중치
- n은 뉴런의 입력 개수



### (4) 정리
1. 동그라미 (뉴런) 의 값과 화살표 (가중치) 의 값을 곱한다
2. 곱한 값을 모두 더한다
3. 여기에 편향을 더한다 --> 결과값 k
4. 그 결과값 k 를 0-1 사이의 값으로 만들어 주는 함수 f(x) 에 전달한다

위 과정에서 결과값 k 를 activatoin 이라고 하고 0-1 사이의 값으로 변환해주는 함수 f(x) 를 활성화 함수라고 한다

앞에서 전제한 뉴런은 그냥 0-1 사이의 어떤 숫자 라 함은 이전 층의 뉴런으로부터 처리된 k 값


## #04. 활성화 함수의 종류

### (1) 선형 함수
- 단층 퍼셉트론
- 선형 함수 외의 모든 활성화 함수는 비선형


- 다중선형회귀 분석에선 대부분 비선형 이다보니 다른 활성화 함수 적용되는 경우 많음

### (2) 시그모이드 함수
- 로지스틱 회귀 분석과 유사
- 0-1 사이의 확률값을 가짐
- 이진분류의 출력층 노드에서 0-1 사이의 값을 만들고 싶을 때 사용한다
- 단점 : input 값이 너무 크거나 작아지면 기울기가 거의 0 이 된다

### (3) 하이퍼볼릭 탄젠트 함수
- 시그모이드 함수의 크기와 위치 조절
- 범위는 [-1,1]
- 그래프의 모양은 0을 기준으로 대칭
- 이 때문에 하이퍼볼릭탄젠트는 시그모이드를 활성화 함수로 썼을 떄보다 학습 수렴 속도가 빠름


### (4) 소프트맥스 함수
- 표준화지수 함수로도 불림
- 출력값이 여러 개로 주어지고 목표치가 다범주인 경우 각 범주에 속할 확률을 제공하는 함수
- 카테고리 분류에 사용


### (5) Relu 함수
- 입력값 x 가 0 이하일떄는 0, 0 초과일 떄는 x 값을 가지는 함수
- 최근 딥러닝에서 많이 활용
- 대부분의 경우 일반적으로 ReLU 의 성능이 가장 좋음
- 어떤 활성화 함수를 사용할 지 모르겠다면 ReLU
- 대부분의 input 값에 대해 기울기가 0이 아니기 때문에 학습이 빨리 된다


## #05 손실 함수의 이해
### [1] 손실함수
- 머신러닝,딥러닝은 결국 컴퓨터가 독립변수로부터 종속변수에 도달하기 위한 가중치와 편향을 찾는 과정


2a + b = 3
3a + b = 7
4a + b =9
9328742946a + b = 18,657,485,893

>> 위와 같은 문제가 있을 때 컴퓨터는 a와 b 를 찾기 위해 값을 직접 대입해 가면서 문제를 품 (노가다처럼) 

>> 이런 삽질을 줄이기 위해 나온게 손실함수

- 대립한 결과값과 실제 정답간의 간격 (오차) 를 최대한 줄이는 방향으로 값을 찾아간다
- 컴퓨터가 선정한 임의의 a,b 를 대입한 결과값이 실제 값에 근사하게 접근하도록 적용하는 특정 함수를 손실함수라고 한다
- 머신러닝이 학습을 통해 찾아낸 a 를 활용하여 구성된 회기식이 실제 정답과의 차이가 거의 없도록 보정하기 위해 다음번에 적용할 a 값을 찾기 위해 사용하는 수학 함수



#### 손실함수의 종류

|--|--|
|회귀분석|MSE,RMSE,MAE|

### [2] 경사 하강법
- 1차 방정식의 근삿값을 발견하기 위한 최적화 알고리즘
- 신경망은 가중치를 업데이트 (a를 끊임없이 바꾸어 가면서 결과값을 예측) 하면서 주어진 문제 최적화
- 가중치를 업데이트하는 가장 대표적인 방법이 경사하강법
- 경사하강법은 특정 함수에서의 미분을 통해 얻은 기울기를 이용하여 최적의 값을 찾아가는 (손실을 줄이는) 방법



- 처음에 값을 랜덤으로 찍어보면서 우리가 찾아야하는 값과 차이가 얼마나 나는지 게속 찾아보는 것 


### Optimizer
- 손실을 최소화 시키면서 경사를 내려가는 방법을 결정하는 알고리즘
- 최적의 Gradient Descent 를 적용하며 최소의 loss 로 빠르고 안정적인 방법을 찾는다


GD → SGD → Momentum → NAG
              ↓
        Adagrad → RMSProp → AdaDelta
                      ↓
                    Adam → Nadam


| 알고리즘 | 핵심 아이디어 | 장점 | 단점 |
|----------|--------------|------|------|
| GD | 전체 데이터를 사용해 한 번에 기울기 계산 | 안정적 수렴 | 속도가 매우 느림 |
| SGD | 일부 데이터(미니배치)로 자주 업데이트 | 빠른 학습 | 진동 발생 가능 |
| Momentum | 이전 이동 방향(관성)을 함께 반영 | 진동 감소, 빠른 수렴 | 하이퍼파라미터(γ) 필요 |
| NAG | 관성 방향으로 먼저 이동 후 기울기 계산 | 더 정확한 방향 예측 | 구현이 약간 복잡 |
| Adagrad | 많이 업데이트된 파라미터 학습률 감소 | 희소 데이터에 강함 | 학습률이 계속 감소 |
| RMSProp | 최근 기울기 중심으로 학습률 조절 | 학습률 감소 문제 완화 | 하이퍼파라미터 필요 |
| AdaDelta | RMSProp 개선, 학습률 자동 조정 | 학습률 설정 불필요 | Adam에 비해 덜 사용 |
| Adam | Momentum + RMSProp 결합 | 빠르고 안정적, 가장 많이 사용 | 메모리 사용량 증가 |
| Nadam | Adam + NAG 결합 | 더 빠른 수렴 가능 | Adam 대비 큰 차이 없음 |
